# <font color='#4B9DA9'>  STM32 Nucleo-64 </font>

##  <font color='#547792'> Popis </font>

Vývojový modul STM32 Nucleo-64 môže byť osadený rôznymi typmi procesorov triedy STM32 v 64-pinovom púzdre LQFP. Na module sú vyvedené konektory pre pripojenie periférií kompatibilných s *Arduino UNO* a konektory s vyvedenými pinmi mikrokontroléra označené ako *ST morpho extension header*. Nižšie uvedený popis zapojenia modulu Nucleo-64 sa vzťahuje k procesoru  [STM32L476](./doc/stm32l476). 

##  <font color='#547792'> Zapojenie </font>

Zapojenie pinov a elektrické zapojenie modulov Nucleo-64 pre rôzne typy mikrokontrolérov sú uvedené v  [dokumentácii](./doc/nucleo_64.pdf). Mapovanie pinov modulu Nocle-64 na konektor Arduin-UNO je v prílohe [pins_L476.csv](./doc/pins_L476.csv).

```{figure} ./img/nucleo-l476.png
:width: 500px
:name: mcp_0060a

Zapojenie pinov modulu STM32 Nucleo-64 s mikrokontrolérom STM32L476.
```

Pri inicializácií periférií môžeme použiť označenie pinov podľa konektorov *ST morpho* alebo konektorov *Arduino*. Periférie mikrokontroléra môžu byť alternatívne pripojené na piny, ktoré môžu mať niekoľko funkcií, tabuľka priradení alternatívnych funkcií je v prílohe [stm32l476_af.csv](./doc/stm32l476_af.csv).

### <font color='#E37434'> Periférne zariadenia </font>


| Periféria         | Nucleo-64 | Arduino-UNO | Poznámka
| :---              | :--       | :--         | :--  
| Zelená LED        | PA5       |             |
| Modré tlačítko    | PC13      |             |


### <font color='#E37434'> Sériové rozhranie I2C </font>

| Periféria         | Nucleo-64 | Arduino-UNO | Poznámka
| :---              | :--       | :--         | :--  
| I2C1 SCL          | PB8       |             |
| I2C1 SDA          | PB9       |             |
| I2C2 SCL          | PB10      |             |
| I2C2 SDA          | PB11      |             |
| I2C3 SCL          | PC0       |             |
| I2C3 SDA          | PC1       |             |


### <font color='#E37434'> Sériové rozhranie SPI </font>

| Periféria         | Nucleo-64 | Arduino-UNO | Alternatíva
| :---              | :--       | :--         | :--  
| SPI1 NSS          | PA4 - NC  | A2          | PA15 - NC
| SPI1 SCK          | PB3       | D3          | PA5
| SPI1 MISO         | PB4       | D5          | PA6
| SPI1 MOSI         | PB5       | D4          | PA7
| SPI2 NSS          | PB12 - NC |             | PB9 - NC
| SPI2 SCK          | PB13      |             | PB10
| SPI2 MISO         | PB14      |             | PC2
| SPI2 MOSI         | PB15      |             | PC3

* NC - Piny NSS sú podporované konfiguráciou STM32, v implementácii *MicroPython-n* nie sú použité


### <font color='#E37434'> Časovače </font>

| Periféria         | Nucleo-64 | Arduino-UNO | Poznámka
| :---              | :--       | :--         | :--  
| TIM2 CH3          | PB10      | D6          |
| TIM3 CH1          | PB4       | D5          |
| TIM3 CH2          | PC7       | D9          |



### <font color='#E37434'> Komunikačné rozhranie USART </font>

| Periféria         | Nucleo-64 | Arduino-UNO | Poznámka
| :---              | :--       | :--         | :--  
| USART1 TX         | PA9       | D8          |
| USART1 RX         | PA10      | D2          |
| USART2 TX         | PA2       | D1          | NC, SB63
| USART2 RX         | PA3       | D0          | NC, SB62
| USART2 RTS        | PA1       | A1          |
| USART2 CTS        | PA0       | A0          |
| USART3 TX         | PC4       |             |  
| USART3 RX         | PC6       |             |  
| USART3 RTS        | PB14      |             |
| USART3 CTS        | PB13      |             |
| USART4 TX         | PA0       | A0          |
| USART4 RX         | PA1       | A1          |
| USART5 TX         | PC12      |             |
| USART5 RX         | PD2       |             |
| LPUART1 TX        | PA12      |             |
| LPUART1 RX        | PA11      |             |    

### <font color='#E37434'> Komunikačné rozhranie CAN  </font>

| Periféria         | Nucleo-64 | Arduino-UNO | Poznámka
| :---              | :--       | :--         | :--  
| CAN1 TX           | PA12      |             |
| CAN1 RX           | PA11      |             |


### <font color='#E37434'> Prevodník ADC </font>

| Periféria         | Nucleo-64 | Arduino-UNO | Poznámka
| :---              | :--       | :--         | :--  
| ADC IN1           | PC0       | A5          |
| ADC IN2           | PC1       | A4          |
| ADC IN3           | PC2       |             |
| ADC IN4           | PC3       |             |


##  <font color='#547792'> Literatúra a odkazy </font>
